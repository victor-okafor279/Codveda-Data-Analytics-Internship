```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
churn_data = pd.read_csv("Cleaned churn-bigml-80.csv")
churn_data.head(5)
```

```python
print("Mean")
print(churn_data.mean(numeric_only=True))
```

```python
print("Median")
print(churn_data.median(numeric_only=True))
```

```python
print("Mode")
print(churn_data.mode())
```

```python
print("Standard Deviation")
print(churn_data.std(numeric_only=True))
```

```python
churn_data.describe()
```

```python
churn_data.hist(figsize=(16,14))

plt.suptitle("Distribution of Numerical Features")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(16,8))

sns.boxplot(data=churn_data.select_dtypes(include=np.number))

plt.xticks(rotation=90)

plt.title("Boxplot of Numerical Features")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Total day minutes",
    y="Total day charge",
    data=churn_data
)

plt.title("Total Day Minutes vs Total Day Charge")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Total eve minutes",
    y="Total eve charge",
    data=churn_data
)

plt.title("Total Evening Minutes vs Total Evening Charge")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Total intl minutes",
    y="Total intl charge",
    data=churn_data
)

plt.title("Total International Minutes vs Total International Charge")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Customer service calls",
    y="Total intl calls",
    data=churn_data
)

plt.title("Customer Service Calls vs International Calls")

plt.show()
```

```python
correlation = churn_data.corr(numeric_only=True)

print(correlation)
```

```python
plt.figure(figsize=(14,10))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="Churn",
    data=churn_data
)

plt.title("Distribution of Customer Churn")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="International plan",
    data=churn_data
)

plt.title("International Plan")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="Voice mail plan",
    data=churn_data
)

plt.title("Voice Mail Plan")

plt.show()
```

```python
# Conclusion

The exploratory data analysis (EDA) of the customer churn dataset provided valuable insights into customer usage patterns and service characteristics. Summary statistics, including the mean, median, mode, and standard deviation, were calculated for all numerical variables to understand their distributions and variability.

Histograms showed the distribution of numerical features, while boxplots helped identify potential outliers. Scatter plots demonstrated strong positive relationships between call minutes and their corresponding charges, indicating that charges increase proportionally with customer usage.

The correlation matrix and heatmap revealed the relationships among the numerical variables, highlighting features that are strongly associated with each other. Count plots also provided an overview of the distribution of customer churn and service plan subscriptions.

Overall, the dataset is clean, well-structured, and suitable for further predictive modeling and customer churn analysis.
```
