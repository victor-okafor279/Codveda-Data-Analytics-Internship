```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sentiment_data = pd.read_csv("Cleaned Sentiment dataset.csv")
sentiment_data.head(5)
```

```python
print("Mean")
print(sentiment_data.mean(numeric_only=True))
```

```python
print("Median")
print(sentiment_data.median(numeric_only=True))
```

```python
print("Mode")
print(sentiment_data.mode())
```

```python
print("Standard Deviation")
print(sentiment_data.std(numeric_only=True))
```

```python
print(sentiment_data.describe())
```

```python
sentiment_data.hist(figsize=(10,8))

plt.suptitle("Distribution of Numerical Features")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=sentiment_data)

plt.title("Boxplot of Numerical Features")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Retweets",
    y="Likes",
    data=sentiment_data
)

plt.title("Retweets vs Likes")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Hour",
    y="Likes",
    data=sentiment_data
)

plt.title("Hour vs Likes")

plt.show()
```

```python
correlation = sentiment_data.corr(numeric_only=True)

print(correlation)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x="Platform", data=sentiment_data)

plt.title("Distribution of Platforms")

plt.xticks(rotation=45)

plt.show()
```

```python
# Conclusion

The exploratory data analysis (EDA) of the Sentiment dataset provided useful insights into both the numerical and categorical variables. Summary statistics, including the mean, median, mode, and standard deviation, were calculated for the numerical features to understand their distributions and variability.

Histograms illustrated the distribution of numerical variables, while boxplots helped identify potential outliers in features such as Retweets and Likes. Scatter plots were used to examine the relationships between Retweets, Likes, and posting Hour, revealing patterns in user engagement.

The correlation matrix and heatmap showed the strength and direction of relationships among the numerical variables. Additionally, the distribution of sentiment categories and social media platforms provided further insight into the dataset.

Overall, the dataset is clean, well-structured, and suitable for further analysis, visualization, and machine learning applications.
```
