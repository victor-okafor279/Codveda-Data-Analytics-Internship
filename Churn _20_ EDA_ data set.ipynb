```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
churn_data = pd.read_csv("Cleaned churn-bigml-20.csv")
churn_data.head(5)
```

```python
print("Mean")
print(churn_data.mean(numeric_only=True))
```

```python
print("Median")
print(churn_data.median(numeric_only=True))
```

```python
print("Mode")
print(churn_data.mode())
```

```python
print("Standard Deviation")
print(churn_data.std(numeric_only=True))
```

```python
churn_data.describe()
```

```python
churn_data.hist(figsize=(16,14))

plt.suptitle("Distribution of Numerical Features")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(15,8))

sns.boxplot(
    data=churn_data.select_dtypes(include=np.number)
)

plt.xticks(rotation=90)

plt.title("Boxplot of Numerical Features")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Total day minutes",
    y="Total day charge",
    data=churn_data
)

plt.title("Total Day Minutes vs Total Day Charge")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Total night minutes",
    y="Total night charge",
    data=churn_data
)

plt.title("Total Night Minutes vs Total Night Charge")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Customer service calls",
    y="Total intl calls",
    data=churn_data
)

plt.title("Customer Service Calls vs International Calls")

plt.show()
```

```python
correlation = churn_data.corr(numeric_only=True)

print(correlation)
```

```python
plt.figure(figsize=(12,10))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="Churn",
    data=churn_data
)

plt.title("Distribution of Churn")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="International plan",
    data=churn_data
)

plt.title("International Plan Distribution")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="Voice mail plan",
    data=churn_data
)

plt.title("Voice Mail Plan Distribution")

plt.show()
```

```python
# Conclusion

The exploratory data analysis (EDA) of the Churn dataset provided valuable insights into customer behavior and service usage. Summary statistics, including the mean, median, mode, and standard deviation, were calculated for all numerical variables to understand their distributions and variability.

Histograms illustrated the distribution of the numerical features, while boxplots helped identify potential outliers. Scatter plots revealed strong positive relationships between usage variables and their corresponding charges, such as total day minutes versus total day charge, indicating that charges increase proportionally with usage.

The correlation matrix and heatmap highlighted the strength and direction of relationships among the numerical variables. Additionally, count plots showed the distribution of customers based on churn status, international plan, and voice mail plan.

Overall, the dataset is clean, well-structured, and suitable for predictive modeling and customer churn analysis.
```
