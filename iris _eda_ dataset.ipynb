```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris_data = pd.read_csv("Cleaned_Iris_Dataset.csv")
iris_data.head(5)
```

```python
print("Mean")
print(iris_data.mean(numeric_only=True))
```

```python
print("Median")
print(iris_data.median(numeric_only=True))
```

```python

print("Mode")
print(iris_data.mode())
```

```python
print("Standard Deviation")
print(iris_data.std(numeric_only=True))
```

```python
print(iris_data.describe())
```

```python
iris_data.hist(figsize=(10,8))

plt.suptitle("Distribution of Numerical Features")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=iris_data)

plt.title("Boxplot of Numerical Features")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="sepal_length",
    y="petal_length",
    hue="species",
    data=iris_data
)

plt.title("Sepal Length vs Petal Length")

plt.show()
```

```python
correlation = iris_data.corr(numeric_only=True)

print(correlation)
```
