```python
import pandas as pd 
import numpy as np
```

```python
# Display Dataset
sentiment_data = pd.read_csv("Sentiment dataset.csv")
sentiment_data.head(10)
```

```python
# Check data types
print(sentiment_data.dtypes)
```

```python
# Check missing values
missing_values = sentiment_data.isnull().sum()
print(missing_values)
```

```python
# Check duplicated rows
num_duplicates = sentiment_data.duplicated().sum()
print("Number of duplicate rows:", num_duplicates)
```

```python
# standardize the time stamp column
sentiment_data["Timestamp"] = pd.to_datetime(sentiment_data["Timestamp"])
```

```python
#check the output to know if standardization is required 
sentiment_data["Sentiment"] = sentiment_data["Sentiment"].str.strip()
```

```python
print(sentiment_data["Platform"].unique())
```

```python
# standardize the platform column to remove extra spaces
sentiment_data["Platform"] = sentiment_data["Platform"].str.strip()
```

```python
# Check the standardize platform column
print(sentiment_data["Platform"].unique())
```

```python
# standardize the country column to remove extra spaces
print(sentiment_data["Country"].unique())
```

```python
sentiment_data["Country"] = sentiment_data["Country"].str.strip()
```

```python
print(sentiment_data["Country"].unique())
```

```python
# check if the Sentiment column has been standardized
print(sentiment_data["Sentiment"].unique())
```

```python
print(sentiment_data["Hashtags"].unique())
```

```python
sentiment_data["Hashtags"] = sentiment_data["Hashtags"].str.strip()
```

```python
print(sentiment_data["Hashtags"].unique())
```

```python
# Remove the unnecessary columns
sentiment_data.drop(columns=["Unnamed: 0", "Unnamed: 0.1"], inplace=True)
```
