```python
import pandas as pd 
import numpy as np
```

```python
# Display Dataset

Stock_data = pd.read_csv("Stock Prices Data Set.csv")
Stock_data.head(5)
```

```python
print(Stock_data.dtypes)
```

```python
# Check for mising values
missing_values = Stock_data.isnull().sum()
print(missing_values)
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy="mean")
Stock_data[["open", "high", "low"]] = imputer.fit_transform(
    Stock_data[["open", "high", "low"]]
)
```

```python
print(Stock_data.isnull().sum())
```

```python
# Check duplicated Values
num_duplicates = Stock_data.duplicated().sum()
print("Number of duplicate rows:", num_duplicates)
```

```python
Stock_data["date"] = pd.to_datetime(Stock_data["date"])
```

```python
print(Stock_data.dtypes)
```

```python
print(Stock_data["symbol"].unique())
```

```python
Stock_data .to_csv("Cleaned Stock Prices Data Set.csv", index=False)
```
