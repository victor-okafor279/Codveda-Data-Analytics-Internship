```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
house_data = pd.read_csv("Cleaned_house_Dataset.csv")
house_data.head(5)
```

```python
# Separate column properly
house_data = pd.read_csv(
    "Cleaned_house_Dataset.csv",
    sep=r"\s+",
    header=None
)
```

```python
# Display Dataset
house_data.head()
```

```python
print("Mean")
print(house_data.mean(numeric_only=True))
```

```python
print("Median")
print(house_data.median(numeric_only=True))
```

```python
print(house_data.mode().iloc[0])
```

```python
print("Standard Deviation")
print(house_data.std(numeric_only=True))
```

```python
print(house_data.describe())
```

```python
house_data.hist(figsize=(18,15))

plt.suptitle("Distribution of Numerical Features")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(15,8))

sns.boxplot(data=house_data)

plt.xticks(rotation=90)

plt.title("Boxplot of Numerical Features")

plt.show()
```

```python
print(house_data.columns)
```

```python
house_data = pd.read_csv(
    "Cleaned_house_Dataset.csv",
    sep=r"\s+",
    header=None
)
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=house_data[0],
    y=house_data[13]
)

plt.xlabel("Column 0")
plt.ylabel("Column 13")
plt.title("Scatter Plot of Column 0 vs Column 13")

plt.show()
```

```python
correlation = house_data.corr(numeric_only=True)

print(correlation)
```
