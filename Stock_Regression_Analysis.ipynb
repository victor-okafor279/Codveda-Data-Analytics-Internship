```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
Stock_data = pd.read_csv("Cleaned Stock Prices Data Set.csv")
Stock_data.head(5)
```

```python
X = Stock_data[["open"]]
Y = Stock_data["close"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
print("Coefficient (Slope):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
mse = mean_squared_error(Y_test, Y_pred)
r2 = r2_score(Y_test, Y_pred)

print("Mean Squared Error:", mse)
print("R-squared:", r2)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(X_test, Y_test, color="blue", label="Actual Data")

plt.plot(X_test, Y_pred, color="red", linewidth=2, label="Regression Line")

plt.title("Simple Linear Regression: Open Price vs Close Price")
plt.xlabel("Opening Price")
plt.ylabel("Closing Price")

plt.legend()

plt.show()
```

```python
comparison = pd.DataFrame({
    "Actual Close Price": Y_test,
    "Predicted Close Price": Y_pred
})

print(comparison.head(10))
```

```python
# Conclusion

A simple linear regression model was developed to predict the closing stock price using the opening stock price.

The dataset was divided into training and testing sets using an 80:20 ratio. The Linear Regression model was trained on the training data and evaluated using the testing data.

The model produced a regression coefficient of **0.999**, indicating that for every one-unit increase in the opening stock price, the predicted closing price increases by approximately **0.999 units**. The intercept was **0.0213**, representing the predicted closing price when the opening price is zero.

The model achieved a **Mean Squared Error (MSE) of 2.65**, indicating that the prediction error is very low. It also produced an **R-squared (R²) value of 0.997**, meaning that approximately **99.7% of the variation in the closing stock price is explained by the opening stock price**.

Overall, the results show that the model performs exceptionally well and that the opening stock price is a very strong predictor of the closing stock price in this dataset.
```
