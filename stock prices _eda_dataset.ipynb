```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Stock_data = pd.read_csv("Cleaned Stock Prices Data Set.csv")
Stock_data.head(5)
```

```python
Stock_data.info()
```

```python
print("Mean")
print(Stock_data.mean(numeric_only=True))
```

```python
print("Median")
print(Stock_data.median(numeric_only=True))
```

```python
print("Mode")
print(Stock_data.mode())
```

```python
print("Standard Deviation")
print(Stock_data.std(numeric_only=True))
```

```python
Stock_data.describe()
```

```python
Stock_data.hist(figsize=(12,10))

plt.suptitle("Distribution of Numerical Features")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=Stock_data[["open","high","low","close","volume"]]
)

plt.title("Boxplot of Numerical Features")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="open",
    y="close",
    data=Stock_data
)

plt.title("Open Price vs Close Price")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="high",
    y="low",
    data=Stock_data
)

plt.title("High Price vs Low Price")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="volume",
    y="close",
    data=Stock_data
)

plt.title("Volume vs Close Price")

plt.show()
```

```python
correlation = Stock_data.corr(numeric_only=True)

print(correlation)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Conclusion

The exploratory data analysis (EDA) of the Stock Prices dataset provided useful insights into the numerical features of the data. Summary statistics, including the mean, median, mode, and standard deviation, were calculated to understand the central tendency and variability of stock prices and trading volume.

Histograms illustrated the distribution of the numerical variables, while boxplots helped identify potential outliers in stock prices and trading volume. Scatter plots revealed relationships between opening and closing prices, high and low prices, and trading volume with closing prices.

The correlation matrix and heatmap highlighted the strength and direction of relationships among the numerical variables. Strong correlations between stock price variables indicate that these features move together and can be valuable for predictive analysis.

Overall, the dataset is clean, well-structured, and suitable for further financial analysis, forecasting, and machine learning applications.
```
